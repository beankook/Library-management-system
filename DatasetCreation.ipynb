```python
# Read the original file
with open("train.txt", "r") as f:
    lines = f.readlines()

# Find midpoint
mid = len(lines) // 2

# Write first half
with open("part1.txt", "w") as f1:
    f1.writelines(lines[:mid])

# Write second half
with open("part2.txt", "w") as f2:
    f2.writelines(lines[mid:])

print("File split into part1.txt and part2.txt")
```

```python
phishing_file = open("phishing_urls.txt", "w")
legitimate_file = open("legitimate_urls.txt", "w")

with open("part2.txt", "r") as f:
    for line in f:
        if line.startswith("phishing"):
            phishing_file.write(line)
        elif line.startswith("legitimate"):
            legitimate_file.write(line)

phishing_file.close()
legitimate_file.close()

print("URLs separated into phishing_urls.txt and legitimate_urls.txt")
```

```python
# Step 1: Combine phishing and legitimate files
with open("combined_urls.txt", "w") as outfile:
    for fname in ["phishing_urls.txt", "legitimate_urls.txt"]:
        with open(fname) as infile:
            outfile.write(infile.read())

# Step 2: Combine with part1.txt
with open("final_dataset.txt", "w") as final:
    for fname in ["part1.txt", "combined_urls.txt"]:
        with open(fname) as infile:
            final.write(infile.read())

print("Files successfully merged into final_dataset.txt")
```
